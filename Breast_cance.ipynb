4 Data preprocessing


```python
# separate data into features and target
X = df.drop(columns=["diagnosis"]) # dropping diagnosis column
y = df["diagnosis"]
```

```python
#train test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=RANDOM_STATE

)
```

```python
print("Dataset size", X.shape)
print("Train Dataset size", X_train.shape)
print("Test Dataset size", X_test.shape)
```

5 Baseline Model - logistic Regression

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
model = LogisticRegression()
```

```python
# training the model
model.fit(X_train_scaled, y_train)
```

```python
# training accuracy
model.score(X_train_scaled, y_train)
```

```python
# test accuracy 
model.score(X_test_scaled, y_test)
```

```python
# model optmisation **
```

```python
# prinicpal component anlaysis best columns
#hyper parameter reduce the recall for false predicition
pipeline = Pipeline(
    [
        ("scaler", StandardScaler()),
        ("pca", PCA()),
       ("model", SVC(kernel="rbf"))

        
    ]
)
```

```python
# grid of hyperparameters
#pca with hyparameter
param_grid = [
 {
   "pca__n_components": [10, 15, 20, 25, 30],
   "model__C": [0.1, 1, 10, 100, 1000],
   "model__gamma": [0.001, 0.01, 0.1, 1, "scale"] 

 }
]
```

```python
# cross-validation setup (used inside grid search cv)
k=5
skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=RANDOM_STATE)
```

```python
# hyperparameter tuning
grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="recall",
    cv=skf,
    n_jobs=-1

)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Hyperparameter Tuning result:")
print("Best params:", grid_search.best_params_)
print("Best params:", grid_search.best_score_)
```

```python
#Train with best params
best_pipeline = Pipeline(
    [
        ("scaler", StandardScaler()),
        ("pca", PCA(n_components=10)),
        ("model", SVC(kernel="rbf", C=10, gamma="scale"))
    ]
)
```

```python
# train the pipeline
best_pipeline.fit(X_train, y_train)
```

8 Model evalution

```python
# training data predicition
y_train_pred = best_pipeline.predict(X_train)
```

```python
train_acc = accuracy_score(y_train, y_train_pred)
print(f"Training Accuracy: {round(train_acc*100, 2)} %")
```

```python
# confusion matrix
cm = confusion_matrix(y_train, y_train_pred)

plt.figure(figsize=(4,3))
#heatmap
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["pred Benign", "Pred malignant"],
    yticklabels=["Actual Benign", "Actual Malignant"]
)
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("confusion matrix heatmap")
plt.show()
```

```python
print("Training Classification report")
print(classification_report(y_train, y_train_pred))
```

```python
#test data prediction
y_test_pred = best_pipeline.predict(X_test)
```

```python
test_acc = accuracy_score(y_test, y_test_pred)
print(f"Test Accuracy: {round(test_acc*100, 2)}")
```

```python
print("Test Classification report")
print(classification_report(y_test, y_test_pred))
```

Build a predictive system

```python
def predict_cancer(input_features):
   input_df = pd.DataFrame(
      [input_features],
      columns=X_train.columns
   )
   prediction = best_pipeline.predict(input_df)
   print("model predcition:", prediction) 
   if prediction[0] == 1:
      print("Diagnosis - Maligant danger")
   else:
      print("Diagnosis - Benign")
      
```
